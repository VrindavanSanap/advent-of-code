```python
with open("./src/2016/day1/2016_day1.txt") as f:
    data = f.read().strip()
```

```python
import numpy as np
```

```python
def find_first_repeat(data):
	directions = [
    [ 0, 1], #north
    [ 1, 0], #east
    [ 0,-1], #south
    [-1, 0], #west
	]
	current_dir = 0
	directions = np.array(directions)
	loc = np.array([0, 0])
	visited = {(0,0)}
	for line in data.split(','):
		line = line.strip()
		turn = line[0]
		dist = line[1:]

		if turn == 'R':
			current_dir += 1
		elif turn == 'L':
			current_dir -= 1
		current_dir = current_dir  % 4
		for i in range(int(dist)):
			loc += directions[current_dir]
			if tuple(loc) in visited:
				return abs(loc).sum()
			visited.add(tuple(loc))
print(find_first_repeat(data))
```
