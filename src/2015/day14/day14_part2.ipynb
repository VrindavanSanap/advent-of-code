```python
with open("./src/2015/day14/2015_day14.txt") as f:
	lines = f.read().split('\n')
```

```python
class Reindeer:
	def __init__(self, speed, run_time, rest_time):
		self.speed = speed
		self.run_time = run_time 
		self.rest_time = rest_time 
		self.cycle_time = run_time + rest_time
		self.dist = 0
		self.time = 0
		self.is_running = True
		self.points = 0

	def step(self):
		if(self.is_running):
			self.dist += self.speed
		



		self.time += 1
		self.time = self.time % self.cycle_time 
		# determine if we are current resting or running
		if (self.time < self.run_time):
			self.is_running =True
		else:
			self.is_running = False

	def give_point(self, max_dist):
		if max_dist <= self.dist:
			self.points += 1
			
	def __repr__(self):
		if self.is_running:
			return f"running at {self.dist}km at {self.speed}km/s with {self.points} points"
		else:
			return f"resting at {self.dist}km with {self.points} points"

class Reindeers:
	def __init__(self):
		self.reindeers = {}

	def add_reindeer(self, name, speed, run_time, rest_time):
		reindeer = Reindeer(speed, run_time, rest_time)
		self.reindeers[name] = reindeer

	def step(self):
		for reindeer in self.reindeers.values():
			reindeer.step()	

	def give_points(self):
		max_dist = 0
		for reindeer in self.reindeers.values():
			if reindeer.dist > max_dist:
				max_dist = reindeer.dist
		
		for reindeer in self.reindeers.values():
			reindeer.give_point(max_dist)

	def winner(self):
		max_points = 0
		for reindeer in self.reindeers.values():
			if reindeer.points > max_points:
				max_points = reindeer.points
	
		return max_points
```

```python
reindeers = Reindeers()
for line in lines:
	line = line.split()
	name = line[0]
	speed = int(line[3])
	run_time = int(line[6])
	rest_time = int(line[-2])
	reindeers.add_reindeer(name, speed, run_time, rest_time)
```

```python
for i in range(2503):
	reindeers.step()
	reindeers.give_points()
```

```python

reindeers.winner()
```
