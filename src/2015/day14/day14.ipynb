```python
with open("./src/2015/day14/2015_day14.txt") as f:
	lines = f.read().split('\n')
```

```python
def distance(speed, run_time, rest_time, time):
	dist = 0
	cycle_time = run_time + rest_time
	num_cycles = time // cycle_time
	dist_per_cycle = speed * run_time
	dist += num_cycles * dist_per_cycle 
	remaining_time = time % cycle_time
	if remaining_time <= run_time:
		dist += remaining_time * speed
	else:
		dist += dist_per_cycle
	return dist
```

```python
max_dist = 0
for line in lines:
	line = line.split()
	speed = int(line[3])
	run_time = int(line[6])
	rest_time = int(line[-2])
	dist = distance(speed, run_time, rest_time, 2503)
	if(dist> max_dist):
		max_dist = dist
print(max_dist)
```
