```python
with open("./src/2015/day16/2015_day16.txt") as f:
	lines = f.read().split("\n")
```

```python
gift_sue = {"children": 3, 
"cats": 7, 
"samoyeds": 2, 
"pomeranians": 3, 
"akitas": 0, 
"vizslas": 0,
"goldfish": 5,
"trees": 3,
"cars": 2,
"perfumes": 1}
```

```python
for line in lines:
	line = line.split()
	sue_no  = line[1].strip(':')
	attr1 = line[2].strip(':')
	attr2 = line[4].strip(':')
	attr3 = line[6].strip(':')
	attr1_value = int(line[3].strip(','))
	attr2_value =  int(line[5].strip(','))
	attr3_value = int(line[7].strip(','))
	attr1_gt= gift_sue[attr1]
	attr2_gt= gift_sue[attr2]
	attr3_gt= gift_sue[attr3]
	if(attr1_value == attr1_gt and attr2_value == attr2_gt and attr3_value == attr3_gt):
		print(sue_no)
		break
```
