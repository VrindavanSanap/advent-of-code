```python
with open("./src/2015/day17/2015_day17.txt") as f:
	data = f.read().strip().split()
	data  = [int(i) for i in data]
```

```python
from itertools import combinations
```

```python
s = 0
for j in range(1,len(data) +1):
	for i in combinations(data, j):
		if(sum(i) == 150):
			s += 1
s
```

```python
n = len(data)
s=0
for mask in range(1 << n):  # 1 << n is the same as 2^n
	subset = []
	for i in range(n):
		if mask & (1 << i):  # check if bit i is set
			subset.append(data[i])
	if(sum(subset))==150:
		s+=1
s
```
