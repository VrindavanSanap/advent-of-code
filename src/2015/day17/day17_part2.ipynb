```python
with open("./src/2015/day17/2015_day17.txt") as f:
	data = f.read().strip().split()
	data  = [int(i) for i in data]
```

```python
from itertools import combinations
```

```python
n_elements = 1000
counter = 0 
for j in range(len(data) +1):
	for i in combinations(data, j):
		if(sum(i) == 150):
			if(len(i)< n_elements) :
				counter = 0
				n_elements = len(i)
			if(len(i) == n_elements):
				counter +=1

counter
```
